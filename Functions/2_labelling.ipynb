```python
import pandas as pd
from openai import OpenAI

secret_key = "sk-MvXt21SGahz1zEvxXFor5w"
client = OpenAI(api_key=secret_key, base_url="https://lite.koboillm.com/v1")
```

```python
def get_sentiment(x, y):
    sentiments = []
    df = pd.read_csv(x)
    df = df[['reviewId', 'content']]

    for content in df['content']:
        response = client.responses.create(
            model="openai/gpt-4.1",
            input=(
                "Kamu adalah pakar analisis sentimen bahasa Indonesia. "
                "Klasifikasikan sentimen komentar berikut.\n\n"
                "Aturan:\n"
                "- Balas HANYA dengan angka: 1, 2, atau 0\n"
                "- 1 = Positif\n"
                "- 2 = Negatif\n"
                "- 0 = Netral\n"
                "- Jangan tambahkan penjelasan apapun\n\n"
                "Komentar: " + str(content)
            )
        )

        outputtxt = response.output_text.strip()

        # Validate output — fallback to 0 (neutral) if LLM returns unexpected value
        if outputtxt not in ["0", "1", "2"]:
            print(f"WARNING: Unexpected output '{outputtxt}' for: {content}")
            outputtxt = "0"

        label = int(outputtxt)
        print(f"content  : {content}")
        print(f"sentiment: {label} ({'Positif' if label == 1 else 'Negatif' if label == 2 else 'Netral'})")
        print("=========")
        sentiments.append(label)

    df['sentiment'] = sentiments
    df.to_excel(f'SENTIMENT_{y}.xlsx', index=False)

    print(f"\n=== COMPLETE: {y} ===\n")
    print(df['sentiment'].value_counts())  # quick summary of label distribution
```
