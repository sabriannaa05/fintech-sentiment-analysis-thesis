```python
import os
import pandas as pd
from openai import OpenAI

client = OpenAI(
    api_key=os.environ["OPENAI_API_KEY"],
    base_url="https://lite.koboillm.com/v1",
)
```

```python
def get_normalized(x,y):
    normalized = []
    file_to_normalized = x
    df = pd.read_csv(file_to_normalized)
    df = df[['reviewId','content']]

    for content in df.content:
        response = client.responses.create(
            model="openai/gpt-5-nano",
            input="Kamu adalah pakar bahasa Indonesia. Rewrite komentar berikut ini dengan bahasa indonesia yang baku, lowecase, hapus tanda baca, dan hapus angka: "+str(content)+""
        )
        outputtxt = response.output_text
        print("origin: "+str(content)+"")
        print("normalized: "+outputtxt+"")
        normalized.append(outputtxt)
        print("=========")

    df['normalized'] = normalized
    df.to_excel('../Normalized_Done/NORMALIZED_'+y+'.xlsx', index=False)

    print("=========")
    print("=========")
    print(y)
    print("=== COMPLETE ===")
    print("=========")
    print("=========")
```

```python
get_normalized("../Dataset/sample.csv","sample")
```
