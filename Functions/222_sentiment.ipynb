```python
import pandas as pd
from openai import OpenAI

secret_key = "YOUR_SECRET_KEY"
client = OpenAI(api_key=secret_key, base_url="https://lite.koboillm.com/v1")
```

```python
def get_sentiment(x, y):
    sentiment = []
    df = pd.read_csv(x)  # ← changed from read_excel
    # df.head(200) ← removed, no limit now
    df = df[['reviewId', 'normalized']]

    for text in df['normalized']:
        response = client.responses.create(
            model="openai/gpt-5-nano",
            input="Kamu adalah pakar bahasa Indonesia. Tentukan nilai sentiment text ini, isi dengan salah satu dari positif, negatif, netral: " + str(text)
        )
        outputtxt = response.output_text.strip().lower()

        # Validate output
        if outputtxt not in ["positif", "negatif", "netral"]:
            print(f"WARNING: unexpected output '{outputtxt}' for: {text}")
            outputtxt = "netral"

        print(f"origin   : {text}")
        print(f"sentiment: {outputtxt}")
        print("=========")
        sentiment.append(outputtxt)

    df['sentiment'] = sentiment
    df.to_csv(f'sentiment_{y}.csv', index=False)  # ← changed from to_excel

    print(f"\n=== COMPLETE: {y} ===")
    print(df['sentiment'].value_counts())
```
